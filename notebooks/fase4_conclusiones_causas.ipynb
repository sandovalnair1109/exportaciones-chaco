## Conclusión — Objetivo #4: causas del descenso 2022-2024 y la recuperación 2026

La caída de las exportaciones de Chaco entre 2022 y 2024, y su posterior
recuperación en 2026, **no se explican por precio — se explican por volumen
físico, y ese volumen está atado a un cambio real en qué exporta la
provincia.**

**Lo que se descartó (se puso a prueba contra la caída 2022-2024 y no dio señal):**
- El precio implícito propio de Chaco no cayó en el período crítico: subió
  +12,4% entre 2022 y 2024 (Fase 4.1-4.2).
- El precio internacional de referencia (proxy: soja cruda argentina) se
  desacopla justo en el año que más importa — cayó -20,7% en 2024 mientras
  Chaco subía +15,1% (Fase 4.4).
- El tipo de cambio real (ITCRM) prácticamente no se movió en la ventana
  2022-2024 (+0,2%), y en la serie completa su correlación con el precio de
  Chaco es débil y de signo contrario al esperado (Fase 4.5).

**Lo que queda fuera de alcance (no aplica como explicación de la caída):**
- Las retenciones no pudieron ser causa de la caída 2022-2024 — no es que
  se hayan probado y no dieran señal, es que la cronología disponible
  (`retenciones_chaco.csv`) arranca en enero de 2025, cuando esa caída
  (PP: 461,7M → 140,8M USD) ya había ocurrido por completo. Solo son
  evaluables contra la recuperación 2025-2026, donde tampoco muestran un
  efecto claro sobre el volumen mensual agregado (Fase 4.8).

**Lo que se confirmó:**
- El volumen exportado explica entre el 87% y el 160% de la variación en
  USD en los cinco tramos analizados (Fase 4.3) — el precio, cuando se
  mueve, muchas veces va en la dirección *contraria* a la del valor total,
  reforzando aún más el peso del volumen.
- Ese volumen no se movió parejo entre rubros: **Productos Primarios (PP)
  se derrumbó -69,5% entre 2022 y 2024** (con el volumen explicando 95,4%
  de esa caída), mientras que **Manufacturas de origen agropecuario (MOA)
  creció +36,1%** en el mismo período, también por volumen. Chaco pasó de
  exportar 89,4% como materia prima cruda en 2022 a 65,4% en 2026 — un
  cambio estructural, no un movimiento parejo de "menos de todo" (Fase 4.6,
  Fase 4.9).

**Lectura integrada:** Chaco no exportó menos porque el mundo pagó menos, ni
porque el peso se movió — exportó menos **producto primario sin procesar**,
compensado solo parcialmente por más manufactura agropecuaria. La causa de
fondo (por qué cayó específicamente el volumen físico de PP — clima,
cosecha, decisiones de siembra, crisis del algodón) queda fuera de lo que
estos datos pueden responder directamente, pero la nota de contexto sobre el
complejo algodonero (`docs/fuentes.md`: crisis sostenida hasta mediados de
2026 pese a retenciones en 0% desde enero 2025) es consistente con parte de
esta caída de PP.

Sobre retenciones: quedan sin evaluar para la caída (desfasaje temporal) y
no concluyentes para la recuperación — la ausencia de señal en el volumen
mensual agregado no equivale a "no importan": hay evidencia indirecta de que
sí influyen en el **momento** de las ventas (el cupo de USD 7.000M en DJVE
agotado en 3 días tras el decreto 682/2025 es la prueba más clara), un
efecto que la granularidad mensual y provincial de estos datos no permite
capturar con precisión. Queda como pregunta abierta para un análisis futuro
con datos de mayor frecuencia (semanal, tipo BCR) y sin agregación
provincial.
