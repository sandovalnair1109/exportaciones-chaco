## Conclusión — Objetivo #4: causas del descenso 2022-2024 y la recuperación 2026

La caída de las exportaciones de Chaco entre 2022 y 2024 (-58,2% en USD) y su
recuperación en 2025-2026 **se explican sobre todo por volumen físico, y ese
volumen se concentra en Productos Primarios (PP)**. El precio contribuyó poco y
no alcanza para explicar la magnitud. Estos datos muestran *qué* se movió, no
*por qué* cayó el volumen.

**Lo que se confirmó:**
- El volumen explica entre el 86% y el 159% de la variación en USD en los cinco
  tramos analizados (Fase 4.3); supera el 100% cuando el precio empujó en sentido
  contrario.
- No se movió parejo entre rubros: **PP cayó -69,5% entre 2022 y 2024** (461,7M →
  140,8M USD), con el volumen explicando 95,4% de esa caída, mientras que **MOA
  creció +36,1%** (51,8M → 70,5M), también por volumen. MOA compensó apenas una
  fracción de la caída (+18,7M frente a -320,9M). PP pasó de 89,4% del total en
  2022 a 65,4% en 2026 (Fases 4.6 y 4.9).
- En la recuperación (H1 2025 → H1 2026, +49,2%), el volumen explica el 86,3%.

**El precio: un aporte menor, con un matiz:**
- El precio implícito promedio subió de 398,2 a 447,6 USD/ton (+12,4%), pero es
  **efecto composición**: +89,8 de mezcla de rubros y -40,4 de precio dentro de
  cada rubro. En PP y MOA el precio bajó (-9,5% y -7,7%): un aporte menor frente
  al volumen (Fases 4.1, 4.2 y 4.9).
- El precio de la soja (proxy nacional) bajó -26,7% entre 2022 y 2024, en el mismo
  sentido pero mucho más que el precio dentro de los rubros de Chaco, y no explica
  la magnitud de la caída (Fase 4.4). Su correlación de niveles con Chaco es 0,86
  en 2002-2025.
- El ITCRM casi no se movió en 2022-2024 (+0,2%), así que queda descartado para
  ese período. Su correlación de niveles con el precio de Chaco es negativa y
  depende de la ventana (-0,64 en 2002-2026; -0,37 en 2015-2025): no es causal
  (Fase 4.5).

**Retenciones: sin evaluar para la caída, no concluyentes para la recuperación:**
- Las retenciones existían antes de 2025, pero la cronología disponible
  (`retenciones_chaco.csv`) arranca el 27 de enero de 2025, cuando la caída de PP
  ya había ocurrido. Con estos datos no se puede evaluar su efecto sobre la caída
  2022-2024.
- Contra la recuperación 2025-2026 no hay señal clara en el volumen mensual
  agregado (Fase 4.8). Eso no equivale a "no importan": hay evidencia indirecta
  de que influyen en el **momento** de las ventas (el cupo de USD 7.000M en DJVE
  se agotó en 3 días tras el decreto 682/2025), un efecto que la granularidad
  mensual y provincial de estos datos no permite capturar.
- El algodón funciona como contraejemplo, no como prueba: con 0% de retención
  desde enero de 2025, el complejo nacional no acompañó la recuperación (Fase
  4.11, cifras nacionales).

**Lectura integrada:** Chaco exportó menos sobre todo porque se redujo el volumen
de productos primarios, y no porque el mundo pagara mucho menos ni porque el
peso se moviera; MOA compensó solo una fracción mínima. La causa de fondo (por
qué cayó el volumen físico de PP: clima, cosecha, decisiones de siembra, crisis
del algodón) queda fuera de lo que estos datos pueden responder.

**Preguntas abiertas:** el efecto de las retenciones sobre la caída 2022-2024
(requiere una cronología que cubra esos años) y su efecto sobre el momento de
las ventas (requiere datos de mayor frecuencia, tipo semanal, y sin agregación
provincial).
