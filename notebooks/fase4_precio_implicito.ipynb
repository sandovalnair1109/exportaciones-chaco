### Qué es (y qué no es) el precio implícito

Es un promedio ponderado de **todos los rubros de Chaco juntos** (PP/MOA/MOI/CyE)
— no el precio internacional de ningún commodity puntual (eso se compara en
la Fase 4.4, contra el precio de la soja). Un cambio en la **composición**
de lo exportado puede mover este número tanto como un cambio real de
precios internacionales — ese matiz se retoma en la Fase 4.6.

### Resultado clave

El precio implícito **no bajó** entre 2022 y 2024 — subió (+15,1% en 2024
respecto a 2023, y por encima del nivel de 2022). La caída de valor total
en ese período (USD 516,6M → USD 216,1M) se explica casi enteramente por
una baja de **volumen** (-63% en toneladas), no de precio. Esto matiza la
hipótesis dominante original — el componente de precio internacional no se
sostiene con este primer corte de los datos propios de Chaco.

> **Nota:** 2026 es un año parcial (6 de 12 meses) — el precio implícito de
> 2026 no es directamente comparable a años completos sin ese matiz.
