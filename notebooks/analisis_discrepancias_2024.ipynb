### Posibles causas de discrepancia

1. **Cobertura temporal incompleta** — los microdatos pueden no incluir todos los meses del año.
2. **Registros confidenciales** — rubros agregados por secreto estadístico que el INDEC imputa de otra forma en los totales.
3. **Diferencias de criterio** — peso neto vs. bruto, ajustes de valuación, correcciones posteriores.
