## Fase 4.11 (opcional) — Nota de contexto: el algodón

**Objetivo:** no es un análisis nuevo — es una mención breve de un dato
que no aparece en la serie de Chaco (que no distingue producto específico,
solo Rubro) pero que sirve como contraejemplo que refuerza la conclusión
de la Fase 4.10.

**Fuentes:** `docs/matriz_de_citas.md` (filas de CAME/Coninagro) y
`docs/fuentes.md`. No requiere script — son cifras ya citadas y validadas
en fases anteriores, a nivel **nacional**, no específico de Chaco (aclarado
en cada cifra, ver matriz).

---

### El dato

El algodón es relevante para Chaco — la provincia concentra ~45% del área
algodonera nacional, aunque dejó de ser la principal provincia productora
(ese lugar lo ocupa hoy Santiago del Estero; fuente: noticiasagropecuarias.com,
nov. 2025 — **fecha de referencia del dato aún sin confirmar**, ver matriz
de citas).

El complejo algodonero nacional tuvo una montaña rusa reciente:
- 2024: +144,4% en USD y +115,1% en toneladas vs. 2023 (CAME/Infobae).
- Reportado nov. 2025: exportaciones -12%, importaciones +119% (Coninagro).
- Mayo 2026: algodón "en rojo" en el Semáforo de Economías Regionales de
  Coninagro — crisis sostenida hasta mediados de 2026.

Esto ocurre pese a que el algodón está en el Anexo I de retenciones al
0% desde enero de 2025 (decreto 38/2025 — confirmado en texto oficial que
el Anexo I es 0% genérico; que el algodón esté en ese anexo es según
prensa, no confirmado directamente en el Anexo mismo).

### Por qué importa para el objetivo #4

Es un **contraejemplo útil, no una prueba**: si las retenciones bajas
fueran el motor principal de la recuperación 2026 (hipótesis que la Fase
4.9/4.10 ya dejó como "no concluyente"), el algodón —con 0% de retención
desde hace más de un año— debería haber acompañado esa recuperación. No
lo hizo. Esto es consistente con la conclusión ya alcanzada: el factor
que explica la recuperación es volumen/composición (Fase 4.6, 4.9), no
retenciones.

**Limitación:** esta es una cifra nacional del complejo algodonero, no de
Chaco específicamente — la serie de Chaco no tiene detalle de producto, así
que no se puede aislar "cuánto algodón exportó Chaco" para confirmar si el
patrón nacional se replica en la provincia. Queda como nota de color, no
como conclusión firme sobre Chaco.
