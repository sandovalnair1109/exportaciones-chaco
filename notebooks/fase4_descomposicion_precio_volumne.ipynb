## Fase 4.3 — Descomposición explícita: precio vs. volumen

**Objetivo:** las Fases 4.1 y 4.2 mostraron de forma descriptiva que el precio
implícito NO cayó entre 2022 y 2024 (subió). Este paso lo formaliza con una
descomposición contable exacta, para poder decir con un número preciso —no
solo "parece que es volumen"— qué parte de cada variación en USD responde a
precio y qué parte a volumen.

**Script:** `src/fase4_descomposicion_precio_volumen.py`

---

### Método

Valor = Precio × Volumen, así que todo cambio de valor se puede separar en
tres partes que suman EXACTO el total (no es una aproximación):

- **Efecto precio** = (P1 - P0) × Q0 — variación de precio, al volumen inicial
- **Efecto volumen** = (Q1 - Q0) × P0 — variación de volumen, al precio inicial
- **Efecto interacción** = (P1-P0)×(Q1-Q0) — la parte que ocurre porque ambos
  cambiaron a la vez

Se analizan 5 tramos: la caída año a año (2022→2023, 2023→2024, y el total
2022→2024), y la recuperación (2024→2025, y 2025→2026 comparando primer
semestre contra primer semestre —mismo criterio que la Fase 1.11— para no
comparar un año completo contra uno parcial).

### Conclusión parcial — Fase 4.3

La descomposición confirma con un número exacto lo que 4.1 y 4.2 ya
insinuaban: **el volumen domina en los 5 tramos analizados**, tanto en la
caída como en la recuperación.

| Período | Δ Total (M USD) | Efecto precio | Efecto volumen |
|---|---|---|---|
| 2022→2023 | -212,8 | 5,6% | **96,6%** |
| 2023→2024 | -87,7 | -52,2% | **132,3%** |
| 2022→2024 (total) | -300,6 | -21,3% | **107,9%** |
| 2024→2025 | +18,9 | -51,6% | **158,8%** |
| 2025→2026 (H1 vs. H1) | +50,1 | 9,6% | **86,3%** |

En dos tramos (2023→2024 y 2024→2025) el precio se mueve *en contra* de la
dirección del total —el precio sube mientras el valor cae, y baja mientras
el valor sube— por eso da porcentajes fuera del rango 0-100%: es la forma
correcta de mostrar que el volumen no solo explica el cambio, sino que tuvo
que compensar un precio que empujaba para el otro lado.

**Esto descarta con evidencia cuantitativa la hipótesis de precio
internacional como causa principal** de la caída 2022-2024 y de la
recuperación 2026, al menos en el precio implícito propio de Chaco. Queda
pendiente contrastar esto contra el ITCRM (Fase 4.5) — el precio implícito
en USD podría estar ocultando un efecto de tipo de cambio real que no se ve
